## 2. Theoretical Background

### 2.1 Activation Maximization

**Activation maximization** is a feature visualization technique that synthesizes input images
which maximally activate specific neurons in a trained neural network. Given a network $f$
with frozen weights, the objective is to find an input image $\mathbf{x}^*$ that maximizes
the activation of a target neuron $a_{l,k}$ at layer $l$, channel $k$:

$$\mathbf{x}^* = \arg\max_{\mathbf{x}} \, a_{l,k}(f(\mathbf{x}))$$

where $a_{l,k}(f(\mathbf{x}))$ represents the spatial mean activation of channel $k$ at layer $l$
when processing input $\mathbf{x}$.

### 2.2 Optimization Formulation

In practice, we solve this optimization problem using gradient-based methods. Starting from
a random initialization $\mathbf{x}_0$ (typically Gaussian noise), we perform gradient ascent:

$$\mathbf{x}_{t+1} = \mathbf{x}_t + \eta \cdot \nabla_{\mathbf{x}} \, a_{l,k}(f(\mathbf{x}_t))$$

where:
- $\eta$ is the learning rate (step size)
- $\nabla_{\mathbf{x}}$ denotes the gradient with respect to input pixels
- We maximize (gradient *ascent*) rather than minimize

### 2.3 Regularization for Human-Interpretable Results

Without constraints, unrestricted optimization often produces high-frequency adversarial
patterns that maximally activate neurons but lack human interpretability. We apply several
regularization techniques to generate natural-looking visualizations:

**1. Total Variation (TV) Loss:** Penalizes spatial discontinuities

$$\mathcal{L}_{TV}(\mathbf{x}) = \sum_{i,j} |x_{i+1,j} - x_{i,j}| + |x_{i,j+1} - x_{i,j}|$$

**2. L2 Regularization:** Prevents pixel magnitude explosion

$$\mathcal{L}_{L2}(\mathbf{x}) = \|\mathbf{x}\|_2^2$$

**3. Transformation Robustness:** Random jitter, rotation, and scaling during optimization

**4. FFT Parameterization:** Optimization in frequency domain (decorrelated color space)

The complete regularized objective becomes:

$$\mathbf{x}^* = \arg\max_{\mathbf{x}} \left[ a_{l,k}(f(\mathbf{x})) - \lambda_{TV} \mathcal{L}_{TV}(\mathbf{x}) - \lambda_{L2} \mathcal{L}_{L2}(\mathbf{x}) \right]$$

### 2.4 InceptionV1 Architecture

InceptionV1 (GoogLeNet; Szegedy et al., 2015) employs **Inception modules** that apply
multiple convolutional filter sizes in parallel, enabling efficient multi-scale feature learning.
The layer naming convention follows:

| Layer Name | Description | Feature Complexity |
|-----------|-------------|-------------------|
| `conv2d1` | Initial convolution | Low-level (edges, colors) |
| `mixed3a/b` | Early Inception modules | Textures, oriented edges |
| `mixed4a/b/c/d/e` | Middle Inception modules | Shapes, patterns, parts |
| `mixed5a/b` | Late Inception modules | Objects, high-level concepts |

We focus on **`mixed4a`**, a transitional layer where features become semantically meaningful
while remaining individually interpretable—ideal for studying mid-level visual representations.

## 4. Methodology

### 4.1 Model Loading

We utilize the pre-trained InceptionV1 model from Lucent's model zoo, which employs **standardized layer naming**
consistent with the Distill.pub convention (`mixed3a`, `mixed4a`, etc.).

````python
"""Load the pre-trained InceptionV1 model from Lucent's model zoo.

The model is loaded with ImageNet pre-trained weights and set to
evaluation mode (frozen batch normalization, dropout disabled).
"""


def load_inceptionv1_model(device: torch.device) -> torch.nn.Module:
    """Load and prepare InceptionV1 model for feature visualization.

    Args:
        device: The torch device (cuda/cpu) to load the model onto.

    Returns:
        torch.nn.Module: The pre-trained InceptionV1 model in eval mode.

    Notes:
        The model uses layer names like 'mixed3a', 'mixed4a', etc.
        which correspond to the Inception module outputs.
    """
    print("Loading InceptionV1 model...")
    model = inceptionv1(pretrained=True)
    model = model.to(device)
    model = model.eval()

    # Freeze all parameters (we only optimize the input image)
    for param in model.parameters():
        param.requires_grad = False

    print("Model loaded successfully!")
    return model


# Load the model
model = load_inceptionv1_model(DEVICE)
````

### 4.3 Target Layer Analysis

We analyze the `mixed4a` layer to determine
how many neurons (channels) it contains.

````python
"""Analyze the mixed4a layer structure and channel count.

mixed4a is an Inception module that concatenates outputs from
multiple parallel convolution branches.
"""


def get_layer_output_channels(
    model: torch.nn.Module,
    layer_name: str,
    input_size: tuple = (1, 3, 224, 224)
) -> int:
    """Get the number of output channels for a specific layer.

    Args:
        model: The PyTorch model.
        layer_name: Name of the layer to analyze.
        input_size: Input tensor size for forward pass.

    Returns:
        int: Number of output channels at the specified layer.

    Notes:
        Uses a forward hook to capture the layer's output shape.
    """
    output_shape = None

    def hook_fn(module, input, output):
        nonlocal output_shape
        output_shape = output.shape

    # Find and register hook on the target layer
    for name, module in model.named_modules():
        if name == layer_name:
            handle = module.register_forward_hook(hook_fn)
            break
    else:
        raise ValueError(f"Layer '{layer_name}' not found in model")

    # Run forward pass with dummy input
    with torch.no_grad():
        dummy_input = torch.randn(input_size).to(DEVICE)
        model(dummy_input)

    handle.remove()
    return output_shape[1]  # Channel dimension


# Analyze mixed4a
TARGET_LAYER = 'mixed4a'
num_channels = get_layer_output_channels(model, TARGET_LAYER)

print(f"Layer: {TARGET_LAYER}")
print(f"Number of channels (neurons): {num_channels}")
print(f"We will visualize the first 10 neurons (indices 0-9)")
````

## 5. Experiments

### 5.1 Single Neuron Visualization

To establish the visualization methodology, we first demonstrate the process
for a single neuron before scaling to multiple neurons.

### The render_vis Function

Lucent's `render_vis` function handles all the optimization details:

```python
render.render_vis(
    model,           # The neural network
    objective,       # What to maximize (e.g., "mixed4a:0")
    param_f=None,    # Image parameterization (default: FFT-based)
    optimizer=None,  # Optimizer (default: Adam)
    transforms=None, # Data augmentation for robustness
    thresholds=(512,), # Optimization steps
    verbose=False,   # Print progress
    show_image=True, # Display result
)
```

The objective string format is `"layer_name:channel_index"`.

````python
"""Demonstrate activation maximization for a single neuron.

This cell visualizes neuron 0 of mixed4a layer using Lucent's
render_vis function with default settings.
"""


def visualize_single_neuron(
    model: torch.nn.Module,
    layer_name: str,
    neuron_idx: int,
    image_size: int = 320,
    num_steps: int = 512,
    show: bool = True
) -> np.ndarray:
    """Generate activation maximization visualization for a single neuron.

    Uses gradient ascent to find an input image that maximally activates
    the specified neuron in the network.

    Args:
        model: Pre-trained neural network in eval mode.
        layer_name: Name of the layer containing the target neuron.
        neuron_idx: Index of the neuron (channel) to visualize.
        image_size: Output image resolution (square).
        num_steps: Number of optimization iterations.
        show: Whether to display the result.

    Returns:
        np.ndarray: The generated visualization image (H, W, 3).

    Notes:
        - Uses FFT-based image parameterization for smooth results
        - Applies standard transformation robustness (jitter, scale, rotate)
        - More steps generally produce clearer features
    """
    # Define the optimization objective
    objective = f"{layer_name}:{neuron_idx}"

    print(f"Generating visualization for {objective}...")
    print(f"  Image size: {image_size}x{image_size}")
    print(f"  Optimization steps: {num_steps}")

    # Define image parameterization using FFT for smooth results
    param_f = lambda: param.image(image_size, fft=True, decorrelate=True)

    # Run activation maximization
    images = render.render_vis(
        model,
        objective,
        param_f=param_f,
        thresholds=(num_steps,),
        show_image=show,
        verbose=False,
    )

    # images is a list of images at different thresholds
    # We take the final result
    result_image = images[0][0]  # Shape: (H, W, 3)

    return result_image


# Visualize neuron 0 of mixed4a
print("="*60)
print("SINGLE NEURON VISUALIZATION DEMO")
print("="*60)
demo_image = visualize_single_neuron(
    model,
    layer_name='mixed4a',
    neuron_idx=0,
    image_size=320,
    num_steps=512,
    show=True
)

# Plot the generated visualization with matplotlib
plt.figure(figsize=(8, 8))
plt.imshow(np.clip(demo_image, 0, 1))
plt.title(
    'Activation Maximization: mixed4a:0\n'
    'Image that maximally activates neuron 0',
    fontsize=12,
    fontweight='bold'
)
plt.axis('off')
plt.tight_layout()
plt.show()
````

#### 5.2.1 Performance Considerations: 320px Resolution

We use **320×320 pixels** for feature visualizations, which provides a good balance
between quality and performance. Here's how resolution affects the visualization process:

#### GPU Memory Requirements

Memory scales quadratically with image size due to feature map dimensions:

$$\text{Memory} \propto \text{(image\_size)}^2 \times \text{channels} \times \text{batch\_size}$$

| Resolution | Approx. GPU Memory | Quality | Recommendation |
|-----------|-------------------|---------|----------------|
| 64×64 | ~0.5 GB | Low | Quick testing only |
| 128×128 | ~1-2 GB | Medium | Fast iteration |
| **320×320** | **~4-6 GB** | **High** | **Recommended default** |
| 512×512 | ~8-12 GB | Very High | Publication quality |
| 1024×1024 | ~20+ GB | Maximum | A100/H100 required |

> **Note**: Actual memory usage depends on model architecture and batch size.
> InceptionV1 is relatively lightweight compared to modern architectures.

#### Compute Time

Time per visualization scales with resolution and optimization steps:

$$\text{Time} \propto \text{(image\_size)}^2 \times \text{num\_steps}$$

| Resolution | 256 steps | 512 steps | 1024 steps |
|-----------|----------|----------|------------|
| 128×128 | ~5s | ~10s | ~20s |
| **320×320** | **~15s** | **~30s** | **~60s** |
| 512×512 | ~30s | ~60s | ~120s |

*(Approximate timings on NVIDIA T4/Colab GPU)*

#### Why 320px?

1. **Sufficient Detail**: Shows fine-grained features (textures, edges)
2. **Colab Compatible**: Fits within free tier GPU memory limits
3. **Reasonable Time**: ~30s per visualization is acceptable for exploration
4. **Matches Data**: Original ImageNet training used 224px; 320px captures similar scale

#### 5.2.2 Batch Visualization Generation

We generate activation maximization images for the first 10 neurons
of the target layer using the optimized pipeline described above.


````python
"""Generate activation maximization visualizations for multiple neurons.

This cell creates visualizations for the first 10 neurons of mixed4a
and stores them for later display.
"""


def generate_neuron_visualizations(
    model: torch.nn.Module,
    layer_name: str,
    neuron_indices: list,
    image_size: int = 320,
    num_steps: int = 512
) -> dict:
    r"""
    Generate activation maximization images for multiple neurons.
    
    This function creates feature visualizations by running gradient ascent
    optimization to find inputs that maximally activate specific neurons.
    
    Resolution & Performance Tradeoffs:
    ----------------------------------
    The `image_size` parameter significantly affects both quality and compute:
    
    - **320×320 (default)**: High quality, ~4-6 GB GPU, ~30s per image
    - **128×128**: Medium quality, ~1-2 GB GPU, ~10s per image (for quick tests)
    - **512×512**: Very high quality, ~8-12 GB GPU, ~60s per image (for publication)
    
    Memory scales as O(n²) where n is image_size. Time scales similarly.
    
    Mathematical Background:
    -----------------------
    For each neuron k at layer l, we optimize:
    
        x* = argmax_x [ mean(A_k(f(x))) - λ_TV * L_TV(x) - λ_L2 * ||x||² ]
    
    where:
    - A_k: Activation map for channel k
    - L_TV: Total variation (smoothness regularization)
    - λ: Regularization weights
    
    The FFT parameterization ensures natural-looking images by operating
    in frequency space, avoiding adversarial high-frequency patterns.
    
    Args:
        model: Pre-trained neural network (frozen weights, eval mode).
        layer_name: Target layer name (e.g., 'mixed4a', 'mixed3b').
        neuron_indices: List of neuron/channel indices to visualize.
        image_size: Output resolution in pixels. Default 320 for high quality.
            Higher values require more GPU memory but show finer details.
        num_steps: Gradient ascent iterations. More steps = cleaner features.
            512 is a good default; use 1024+ for publication quality.
    
    Returns:
        dict: Mapping from neuron index to numpy array of shape (H, W, 3)
            with pixel values in range [0, 1].
    
    Raises:
        RuntimeError: If CUDA out of memory (try reducing image_size).
    
    Example:
        >>> # Standard high-quality visualization
        >>> images = generate_neuron_visualizations(
        ...     model, 'mixed4a', range(10),
        ...     image_size=320, num_steps=512
        ... )
        >>> images[0].shape  # (320, 320, 3)
        
        >>> # Quick preview (lower quality, faster)
        >>> quick_images = generate_neuron_visualizations(
        ...     model, 'mixed4a', range(10),
        ...     image_size=128, num_steps=256
        ... )
    
    Performance Notes:
        - On Colab T4 GPU: ~30s per 320×320 image with 512 steps
        - On Colab free tier: 320×320 should work; reduce if OOM errors
        - For 10 neurons: expect ~5 minutes total runtime
    
    References:
        Mordvintsev et al., "Differentiable Image Parameterizations", Distill 2018
        Olah et al., "Feature Visualization", Distill 2017
    """
    visualizations = {}

    # Define image parameterization once (reused for each neuron)
    param_f = lambda: param.image(image_size, fft=True, decorrelate=True)

    total = len(neuron_indices)
    for i, neuron_idx in enumerate(neuron_indices):
        objective = f"{layer_name}:{neuron_idx}"
        print(f"[{i+1}/{total}] Generating {objective}...", end=" ")

        # Generate visualization (suppress display)
        images = render.render_vis(
            model,
            objective,
            param_f=param_f,
            thresholds=(num_steps,),
            show_image=False,
            verbose=False,
        )

        visualizations[neuron_idx] = images[0][0]
        print("Done!")

    return visualizations


# Generate visualizations for first 10 neurons
print("="*60)
print("GENERATING VISUALIZATIONS FOR FIRST 10 NEURONS OF mixed4a")
print("="*60)
print(f"Layer: mixed4a")
print(f"Neurons: 0-9 (10 total)")
print(f"Image size: 320x320")
print(f"Optimization steps: 512")
print("-"*60)

neuron_visualizations = generate_neuron_visualizations(
    model=model,
    layer_name='mixed4a',
    neuron_indices=list(range(10)),
    image_size=320,
    num_steps=512
)

print("-"*60)
print(f"Generated {len(neuron_visualizations)} visualizations!")
````

````python
"""Display neuron visualizations in a publication-quality grid layout.

Creates a 2x5 grid of activation maximization images with labels,
similar to the Distill.pub Circuits visualization style.
"""


def display_neuron_grid(
    visualizations: dict,
    layer_name: str,
    ncols: int = 5,
    figsize: tuple = (15, 7),
    title: Optional[str] = None
) -> plt.Figure:
    """Display neuron visualizations in a grid layout.

    Args:
        visualizations: Dict mapping neuron indices to images.
        layer_name: Name of the layer for labeling.
        ncols: Number of columns in the grid.
        figsize: Figure size (width, height).
        title: Optional title for the entire figure.

    Returns:
        plt.Figure: The matplotlib figure object.
    """
    # Calculate grid dimensions
    n_images = len(visualizations)
    nrows = (n_images + ncols - 1) // ncols

    # Create figure with constrained layout
    fig, axes = plt.subplots(
        nrows, ncols,
        figsize=figsize,
        constrained_layout=True
    )

    # Flatten axes for easy iteration
    if nrows == 1:
        axes = axes.reshape(1, -1)
    axes_flat = axes.flatten()

    # Plot each visualization
    sorted_indices = sorted(visualizations.keys())
    for ax_idx, neuron_idx in enumerate(sorted_indices):
        ax = axes_flat[ax_idx]
        image = visualizations[neuron_idx]

        # Ensure image is in valid range [0, 1] for display
        image_clipped = np.clip(image, 0, 1)

        ax.imshow(image_clipped)
        ax.set_title(f"{layer_name}:{neuron_idx}", fontsize=11, fontweight='bold')
        ax.axis('off')

    # Hide unused subplots
    for ax_idx in range(len(sorted_indices), len(axes_flat)):
        axes_flat[ax_idx].axis('off')
        axes_flat[ax_idx].set_visible(False)

    # Add main title if provided
    if title:
        fig.suptitle(title, fontsize=14, fontweight='bold', y=1.02)

    return fig


# Display the grid
fig = display_neuron_grid(
    neuron_visualizations,
    layer_name='mixed4a',
    ncols=5,
    figsize=(16, 7),
    title='Activation Maximization: First 10 Neurons of InceptionV1 mixed4a'
)

plt.show()
````

### 5.3 Advanced Optimization Techniques

The optimization process can be customized through various parameters
to generate visualizations with different characteristics. Key parameters include:

### Image Parameterization

- **FFT (Fourier) parameterization**: Produces smoother, more natural-looking images
- **Pixel parameterization**: Direct optimization in pixel space (often noisier)
- **Decorrelation**: Optimizes in a decorrelated color space

### Transformations

Random transformations during optimization improve **robustness**:
- **Jitter**: Small random translations
- **Scale**: Random zoom in/out
- **Rotate**: Small random rotations

````python
"""Demonstrate different parameterization options.

Compare FFT-based vs pixel-based parameterization to show
the importance of proper image parameterization.
"""


def compare_parameterizations(
    model: torch.nn.Module,
    layer_name: str,
    neuron_idx: int,
    image_size: int = 320,
    num_steps: int = 256
) -> tuple:
    """Compare FFT vs pixel parameterization for the same neuron.

    Args:
        model: The neural network model.
        layer_name: Target layer name.
        neuron_idx: Neuron index to visualize.
        image_size: Output image resolution.
        num_steps: Optimization iterations.

    Returns:
        tuple: (fft_image, pixel_image) as numpy arrays.
    """
    objective = f"{layer_name}:{neuron_idx}"

    # FFT parameterization (smooth)
    print(f"Generating with FFT parameterization...")
    param_fft = lambda: param.image(image_size, fft=True, decorrelate=True)
    images_fft = render.render_vis(
        model, objective,
        param_f=param_fft,
        thresholds=(num_steps,),
        show_image=False,
        verbose=False
    )

    # Pixel parameterization (often noisier)
    print(f"Generating with pixel parameterization...")
    param_pixel = lambda: param.image(image_size, fft=False, decorrelate=False)
    images_pixel = render.render_vis(
        model, objective,
        param_f=param_pixel,
        thresholds=(num_steps,),
        show_image=False,
        verbose=False
    )

    return images_fft[0][0], images_pixel[0][0]


# Compare for neuron 5
print("Comparing parameterization methods for mixed4a:5")
print("-" * 50)
fft_img, pixel_img = compare_parameterizations(
    model, 'mixed4a', 5, image_size=320, num_steps=256
)

# Display comparison
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(np.clip(fft_img, 0, 1))
axes[0].set_title('FFT Parameterization\n(Smoother, more natural)', fontsize=12)
axes[0].axis('off')

axes[1].imshow(np.clip(pixel_img, 0, 1))
axes[1].set_title('Pixel Parameterization\n(More high-frequency noise)', fontsize=12)
axes[1].axis('off')

fig.suptitle('Effect of Image Parameterization on Feature Visualization',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

### 5.5 Interactive Feature Circuit Visualization

### Theoretical Foundation: Compositional Feature Hierarchies

The **Circuits** research thread from Distill.pub (Olah et al., 2020) demonstrates that
neural networks learn *compositional representations* where features in later layers
are built from combinations of features in earlier layers.

#### Mathematical Framework

For a layer $l$, the activation of neuron $k$ can be expressed as:

$$a_{l,k}(\mathbf{x}) = \sigma\left(\sum_{j} w_{jk}^{(l)} \cdot a_{l-1,j}(\mathbf{x}) + b_k^{(l)}\right)$$

where:
- $a_{l,k}$ = Activation of neuron $k$ at layer $l$
- $w_{jk}^{(l)}$ = Weight connecting neuron $j$ (layer $l-1$) to neuron $k$ (layer $l$)
- $\sigma$ = Non-linear activation function (ReLU)
- $b_k^{(l)}$ = Bias term

#### The Polysemantic Neuron Hypothesis

Individual neurons often respond to multiple, seemingly unrelated concepts.
However, when examined as **circuits** (connected groups of neurons), they reveal
interpretable algorithms.

### Feature Hierarchy in InceptionV1

| Layer | Receptive Field | Feature Complexity | Typical Detections |
|-------|----------------|-------------------|-------------------|
| `conv2d1` | 7×7 | Low | Edges, colors, gradients |
| `mixed3a` | ~35×35 | Medium-Low | Gabor filters, textures |
| `mixed3b` | ~55×55 | Medium | Complex textures, curves |
| `mixed4a` | ~107×107 | Medium-High | Object parts, shapes |
| `mixed5b` | ~224×224 | High | Full objects, scenes |

### Circuit Example: The Car Detector

From the Distill.pub [Zoom In](https://distill.pub/2020/circuits/zoom-in/) article:

```
Windows (4b:237)  ─┐
                   ├──→  Car Detector (4c:447)
Car Body (4b:491) ─┤
                   │
Wheels (4b:373)  ──┘
```

Each component neuron **excites** or **inhibits** the target detector based on
spatial relationships (e.g., wheels excite the bottom of the car detector).

### Visualization Goals

1. **Plotly Grid**: Interactive hover showing feature descriptions
2. **Distill-Style Diagram**: Visual flow from early to late layers

**References:**
- Olah et al., *Zoom In: An Introduction to Circuits*, Distill 2020
- Olah et al., *An Overview of Early Vision*, Distill 2020


````python
r"""
Generate feature visualizations across multiple layers for circuit analysis.

This module creates activation maximization images for neurons at different
depths in InceptionV1, demonstrating how features compose hierarchically.

Mathematical Background:
------------------------
For each neuron, we find the input that maximizes its activation:

    x* = argmax_x mean(A_k(f(x)))

where A_k is the activation map for channel k.

Layer Selection Rationale:
-------------------------
- conv2d1: First conv layer, learns edge/color primitives
- mixed3a: First Inception module, Gabor-like textures
- mixed3b: Second Inception, complex texture compositions
- mixed4a: Third Inception, object parts emerge

References:
    Olah et al., "Zoom In: An Introduction to Circuits", Distill 2020
    https://distill.pub/2020/circuits/zoom-in/
"""

# Layer configuration for circuit analysis
# Each layer represents a different level of the feature hierarchy
CIRCUIT_LAYERS = {
    'conv2d1': {
        'indices': [0, 1, 2, 3],
        'desc': 'Early vision: basic edge and color detectors',
        'receptive_field': '7×7 pixels'
    },
    'mixed3a': {
        'indices': [0, 1, 2, 3],
        'desc': 'Gabor-like oriented edge detectors',
        'receptive_field': '~35×35 pixels'
    },
    'mixed3b': {
        'indices': [0, 1, 2, 3],
        'desc': 'Complex texture and curve detectors',
        'receptive_field': '~55×55 pixels'
    },
    'mixed4a': {
        'indices': [0, 1, 2, 3],
        'desc': 'Object parts: boundaries, shapes, parts',
        'receptive_field': '~107×107 pixels'
    },
}

# Feature metadata: human-interpretable descriptions for each neuron
# Based on visual inspection and Distill.pub feature families
FEATURE_METADATA = {
    # Early vision (conv2d1) - Corresponds to V1-like features
    'conv2d1:0': 'Low-frequency detector: responds to smooth gradients and large-scale brightness changes',
    'conv2d1:1': 'High-frequency detector: responds to fine textures and sharp transitions',
    'conv2d1:2': 'Color opponent (R-G): red-green channel, similar to retinal ganglion cells',
    'conv2d1:3': 'Color opponent (B-Y): blue-yellow channel, fundamental color processing',
    
    # Gabor-like (mixed3a) - Similar to V1 simple cells
    'mixed3a:0': 'Horizontal edge: responds to horizontal oriented boundaries (0°)',
    'mixed3a:1': 'Diagonal edge (45°): responds to edges at 45° orientation',
    'mixed3a:2': 'Vertical edge: responds to vertical oriented boundaries (90°)',
    'mixed3a:3': 'Texture detector: responds to regular repeating patterns',
    
    # Complex textures (mixed3b) - V2/V4-like features
    'mixed3b:0': 'Curve detector: responds to curved contours and arcs',
    'mixed3b:1': 'Corner/junction: responds to L-shapes and T-junctions',
    'mixed3b:2': 'Grid pattern: responds to regular lattice structures',
    'mixed3b:3': 'Organic texture: responds to natural textures (fur, foliage)',
    
    # Object parts (mixed4a) - IT-like features
    'mixed4a:0': 'Silhouette boundary: responds to object outlines against backgrounds',
    'mixed4a:1': 'Window-like: responds to rectangular frames and window patterns',
    'mixed4a:2': 'Circular/wheel: responds to round objects and wheel-like shapes',
    'mixed4a:3': 'Face parts: responds to eye-like or facial feature patterns',
}


def generate_circuit_visualizations(
    model: torch.nn.Module,
    layers: dict,
    image_size: int = 128,
    num_steps: int = 256
) -> dict:
    r"""
    Generate activation maximization images for neurons across multiple layers.
    
    This function creates visualizations that reveal what each neuron has learned
    to detect, enabling analysis of the compositional feature hierarchy.
    
    Mathematical Background:
        The optimization finds x* such that:
        
        x* = argmax_x [ mean(A_k(f(x))) - λ_TV * L_TV(x) - λ_L2 * ||x||² ]
        
        where A_k is the k-th channel activation and L_TV is total variation.
    
    Args:
        model: Pre-trained InceptionV1 model (frozen weights)
        layers: Dict mapping layer names to config dicts with 'indices' key
        image_size: Output image resolution (default: 96 for speed)
        num_steps: Gradient ascent iterations (more = cleaner features)
    
    Returns:
        dict: Mapping 'layer:neuron_idx' -> numpy array (H, W, 3)
    
    Example:
        >>> images = generate_circuit_visualizations(model, CIRCUIT_LAYERS)
        >>> images['mixed4a:0'].shape  # (96, 96, 3)
    
    Notes:
        - Uses FFT parameterization for natural-looking images
        - Decorrelated color space prevents adversarial patterns
        - Lower num_steps trades quality for speed
    
    References:
        Mordvintsev et al., "Differentiable Image Parameterizations", Distill 2018
    """
    circuit_images = {}
    
    # FFT parameterization with decorrelated colors
    # This produces more natural, interpretable visualizations
    param_f = lambda: param.image(image_size, fft=True, decorrelate=True)
    
    total_neurons = sum(len(cfg['indices']) for cfg in layers.values())
    current = 0
    
    for layer_name, config in layers.items():
        print(f"\n📊 Layer: {layer_name}")
        print(f"   Description: {config['desc']}")
        print(f"   Receptive field: {config.get('receptive_field', 'N/A')}")
        
        for idx in config['indices']:
            current += 1
            key = f"{layer_name}:{idx}"
            print(f"   [{current}/{total_neurons}] {key} ... ", end="", flush=True)
            
            try:
                # Run activation maximization
                images = render.render_vis(
                    model, key,
                    param_f=param_f,
                    thresholds=(num_steps,),
                    show_image=False,
                    verbose=False
                )
                circuit_images[key] = images[0][0]
                print("✓")
            except Exception as e:
                print(f"✗ (Error: {str(e)[:50]})")
                # Return blank image on failure
                circuit_images[key] = np.zeros((image_size, image_size, 3))
    
    return circuit_images


# Generate circuit visualizations
print("="*70)
print("GENERATING FEATURE CIRCUIT VISUALIZATIONS")
print("="*70)
print("\nThis generates activation maximization images for neurons across")
print("multiple layers, revealing the compositional feature hierarchy.")
print("\nLayers: conv2d1 → mixed3a → mixed3b → mixed4a")
print("-"*70)

circuit_images = generate_circuit_visualizations(
    model, CIRCUIT_LAYERS,
    image_size=320,   # Smaller for speed; use 128+ for publication quality
    num_steps=256    # Fewer steps for speed; use 512+ for cleaner features
)

print("\n" + "="*70)
print(f"✅ Generated {len(circuit_images)} feature visualizations")
print("="*70)
````

````python
r"""
Create interactive Plotly visualization with hover tooltips.

This visualization allows exploring the feature hierarchy interactively.
Hovering over each feature image displays:
- Neuron identifier (layer:index)
- Human-interpretable description
- Feature family classification

Implementation Notes:
--------------------
Plotly is chosen for Jupyter/Colab compatibility without additional
JavaScript dependencies. The go.Image trace with custom hovertemplate
provides native hover interactivity.

References:
    Plotly documentation: https://plotly.com/python/hover-text-and-formatting/
"""

import plotly.graph_objects as go
from plotly.subplots import make_subplots


def create_plotly_circuit_grid(
    images: dict,
    metadata: dict,
    layers: dict,
    title: str = "Feature Circuit Hierarchy"
) -> go.Figure:
    r"""
    Create an interactive Plotly grid visualization of feature circuits.
    
    The grid is organized with:
    - Columns: Layers (early → late, left → right)
    - Rows: Different neurons within each layer
    
    Args:
        images: Dict mapping 'layer:idx' to numpy image arrays
        metadata: Dict mapping 'layer:idx' to description strings
        layers: Dict of layer configurations
        title: Title for the figure
    
    Returns:
        go.Figure: Interactive Plotly figure with hover tooltips
    
    Example:
        >>> fig = create_plotly_circuit_grid(images, metadata, layers)
        >>> fig.show()  # Opens interactive visualization
    """
    layer_names = list(layers.keys())
    n_layers = len(layer_names)
    n_neurons = max(len(cfg['indices']) for cfg in layers.values())
    
    # Create subplot grid: columns=layers, rows=neurons
    fig = make_subplots(
        rows=n_neurons, cols=n_layers,
        subplot_titles=[f"<b>{ln}</b><br><sup>{layers[ln]['desc'][:30]}...</sup>" 
                       for ln in layer_names],
        horizontal_spacing=0.01,
        vertical_spacing=0.02,
    )
    
    for col, layer_name in enumerate(layer_names, 1):
        for row, idx in enumerate(layers[layer_name]['indices'], 1):
            key = f"{layer_name}:{idx}"
            img = images.get(key)
            
            if img is not None:
                # Convert to uint8 for display
                img_uint8 = (np.clip(img, 0, 1) * 255).astype(np.uint8)
                desc = metadata.get(key, 'No description available')
                
                fig.add_trace(
                    go.Image(
                        z=img_uint8,
                        hovertemplate=(
                            f"<b style='font-size:14px'>{key}</b><br><br>"
                            f"{desc}<br>"
                            f"<extra></extra>"
                        ),
                    ),
                    row=row, col=col
                )
    
    # Update layout for clean appearance
    fig.update_layout(
        title={
            'text': f'<b>{title}</b><br><sup>Hover over images for feature descriptions</sup>',
            'x': 0.5,
            'font': {'size': 18}
        },
        height=1400,
        width=2000,
        showlegend=False,
        paper_bgcolor='#f8f9fa',
        plot_bgcolor='#f8f9fa',
    )
    
    # Hide axes for cleaner look
    fig.update_xaxes(showticklabels=False, showgrid=False, zeroline=False)
    fig.update_yaxes(showticklabels=False, showgrid=False, zeroline=False)
    
    return fig


# Create and display the interactive Plotly grid
print("Creating interactive Plotly visualization...")
plotly_fig = create_plotly_circuit_grid(
    circuit_images,
    FEATURE_METADATA,
    CIRCUIT_LAYERS,
    title="Feature Circuit: conv2d1 → mixed3a → mixed3b → mixed4a"
)
plotly_fig.show()
print("\n💡 Tip: Hover over each image to see what the neuron detects!")
````
